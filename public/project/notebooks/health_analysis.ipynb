```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
sns.set_theme(style='whitegrid')
```

## 1. Data Collection & Understanding

```python
df = pd.read_csv('../data/health_dataset.csv')
print(df.shape)
df.head()
```

```python
df.info()
df.describe()
```

## 2. Data Cleaning

```python
print('Missing values:\n', df.isna().sum())
df = df.drop_duplicates()
print('Shape after dedup:', df.shape)
```

## 3. Exploratory Data Analysis

```python
sns.histplot(df['bmi'], kde=True); plt.title('BMI Distribution'); plt.show()
sns.countplot(x='healthy_lifestyle', data=df); plt.show()
plt.figure(figsize=(9,6)); sns.heatmap(df.select_dtypes('number').corr(), cmap='coolwarm'); plt.show()
```

## 4. Feature Engineering & Preprocessing

```python
d = df.copy()
d['gender'] = d['gender'].map({'Male':0,'Female':1})
d['smoker'] = d['smoker'].map({'No':0,'Yes':1})
d['healthy_lifestyle'] = d['healthy_lifestyle'].map({'No':0,'Yes':1})
X = d.drop(columns=['id','healthy_lifestyle'])
y = d['healthy_lifestyle']
X = StandardScaler().fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)
```

## 5. Model Training & Evaluation

```python
for name, model in [('LogReg', LogisticRegression(max_iter=1000)), ('RF', RandomForestClassifier(random_state=42))]:
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    print(name, 'accuracy:', accuracy_score(y_test, pred))
    print(classification_report(y_test, pred))
```
